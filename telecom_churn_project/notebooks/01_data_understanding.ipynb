# Environment Setup + Dataset Understanding

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
df = pd.read_csv('../data/TelcoCustomerChurn.csv')
print("Shape:", df.shape)
df.head()
```

```python
print("=== COLUMN TYPES ===")
print(df.dtypes)
print("\n=== MISSING VALUES ===")
print(df.isnull().sum())
```

```python
print("=== NUMERICAL SUMMARY ===")
print(df.describe())
print("\n=== CATEGORICAL COLUMNS ===")
cat_cols = df.select_dtypes(include='object').columns.tolist()
for col in cat_cols:
    print(f"\n{col}: {df[col].nunique()} unique → {df[col].unique()}")
```

```python
churn_counts = df['Churn'].value_counts()
churn_pct = df['Churn'].value_counts(normalize=True) * 100

print("=== CHURN DISTRIBUTION ===")
print(pd.DataFrame({'Count': churn_counts, 'Percentage': churn_pct.round(2)}))
```

```python
data_dict = {
    'customerID':       ('Categorical', 'Unique customer identifier — drop before modeling'),
    'gender':           ('Categorical', 'Male / Female'),
    'SeniorCitizen':    ('Binary Int',  '1 = Senior, 0 = Not Senior'),
    'Partner':          ('Categorical', 'Has a partner: Yes/No'),
    'Dependents':       ('Categorical', 'Has dependents: Yes/No'),
    'tenure':           ('Numerical',   'Months with the company'),
    'PhoneService':     ('Categorical', 'Has phone service: Yes/No'),
    'MultipleLines':    ('Categorical', 'Multiple lines: Yes/No/No phone service'),
    'InternetService':  ('Categorical', 'DSL / Fiber optic / No'),
    'OnlineSecurity':   ('Categorical', 'Yes/No/No internet service'),
    'OnlineBackup':     ('Categorical', 'Yes/No/No internet service'),
    'DeviceProtection': ('Categorical', 'Yes/No/No internet service'),
    'TechSupport':      ('Categorical', 'Yes/No/No internet service'),
    'StreamingTV':      ('Categorical', 'Yes/No/No internet service'),
    'StreamingMovies':  ('Categorical', 'Yes/No/No internet service'),
    'Contract':         ('Categorical', 'Month-to-month / One year / Two year'),
    'PaperlessBilling': ('Categorical', 'Yes/No'),
    'PaymentMethod':    ('Categorical', '4 payment types'),
    'MonthlyCharges':   ('Numerical',   'Monthly charge amount in USD'),
    'TotalCharges':     ('String→Float','Total charges — needs type conversion'),
    'Churn':            ('Target',      'Yes = churned, No = retained')
}

dd_df = pd.DataFrame.from_dict(
    data_dict, orient='index', columns=['Type', 'Description']
)
dd_df.index.name = 'Feature'
print(dd_df.to_string())
```

```python
print("=== ISSUES FOUND ===")

# Issue 1: TotalCharges stored as string
print(f"1. TotalCharges dtype: {df['TotalCharges'].dtype} → needs conversion to float")

# Issue 2: Spaces in TotalCharges acting as missing values
tc_spaces = df[df['TotalCharges'].str.strip() == ''].shape[0]
print(f"2. TotalCharges empty strings (hidden nulls): {tc_spaces}")

# Issue 3: Class imbalance
no_pct  = (df['Churn'] == 'No').mean()  * 100
yes_pct = (df['Churn'] == 'Yes').mean() * 100
print(f"3. Class imbalance → No: {no_pct:.1f}%  Yes: {yes_pct:.1f}%")

# Issue 4: customerID is non-predictive
print(f"4. customerID must be dropped before modeling")

# Issue 5: SeniorCitizen encoded as int but is categorical
print(f"5. SeniorCitizen is int but should be treated as binary categorical")
```
